```python
# Q1: Function that takes a string and returns reversed string & every 2nd character
def process_string(s):
    reversed_str = s[::-1]
    every_second = s[1::2]
    return reversed_str, every_second

# Test the function
text = "ArtificialIntelligence"
rev, sec = process_string(text)
print("Original:", text)
print("Reversed:", rev)
print("Every 2nd character:", sec)
```

```python
# Q2: Return squares of even numbers using list comprehension
def square_even_numbers(numbers):
    return [n ** 2 for n in numbers if n % 2 == 0]

# Test the function
nums_list = [1, 2, 3, 4, 5, 6, 7, 8]
print("Original List:", nums_list)
print("Squares of Evens:", square_even_numbers(nums_list))
```

```python
# Q3: Count frequency of each character in a string using a dictionary
def char_frequency(s):
    freq = {}
    for char in s:
        freq[char] = freq.get(char, 0) + 1
    return freq

# Test the function
sample_str = "programming"
print("Character Frequency:", char_frequency(sample_str))
```

```python
# Q4: Take two lists and return a dictionary using zip()
def lists_to_dict(list1, list2):
    return dict(zip(list1, list2))

# Test the function
keys = ['name', 'course', 'semester']
values = ['Haseeba', 'Programming for AI', 4]
print("Combined Dictionary:", lists_to_dict(keys, values))
```

```python
# Q5: Manually iterate through a list using iter(), next(), and try-except for StopIteration
def manual_iterate(my_list):
    iterator = iter(my_list)
    print("Starting manual iteration:")
    while True:
        try:
            element = next(iterator)
            print(element)
        except StopIteration:
            print("StopIteration caught! Reached the end of the list.")
            break

# Test the function
manual_iterate([10, 20, 30, 40])
```

```python
# Q6: Check whether a string is a palindrome using slicing
def is_palindrome(s):
    # Remove spaces and convert to lowercase for accurate check
    cleaned = s.replace(" ", "").lower()
    return cleaned == cleaned[::-1]

# Test the function
print("Is 'Madam' a palindrome?", is_palindrome("Madam"))
print("Is 'Python' a palindrome?", is_palindrome("Python"))
```
